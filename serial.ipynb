```python
import time
import random
```

```python
def gerar_matriz(linhas, colunas):
    return [[random.randint(1, 10) for _ in range(colunas)] for _ in range(linhas)]
```

```python
def multiplicar_matrizes(A, B):
    qtd_linhas = len(A)
    qtd_colunas = len(A[0])
    qtd_colunas_b = len(B[0])
    C = [[0 for _ in range(qtd_colunas_b)] for _ in range(qtd_linhas)] #matriz resultado inicializada com zeros
    for i in range(qtd_linhas):
        for j in range(qtd_colunas_b):
            for k in range(qtd_colunas):
                C[i][j] += A[i][k] * B[k][j] 
    return C   #retorna a matriz resultado da multiplicação de A e B
```
